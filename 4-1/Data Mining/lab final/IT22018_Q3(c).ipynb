Let's break down what each of these metrics represents in the context of association rules, often used in market basket analysis:

```python
import math

total = 5000
hot = 3000
spag = 2500
both = 2000

support = both / total

confidence = both / hot

lift = confidence / (spag / total)


cosine = both / math.sqrt(hot * spag)

print(f"Support: {support:.4f}")
print(f"Confidence: {confidence:.4f}")
print(f"Lift: {lift:.4f}")
print(f"Cosine: {cosine:.4f}")
```
