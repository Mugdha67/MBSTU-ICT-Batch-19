```python
import numpy as np
from sklearn.cluster import KMeans

# Dataset
X = np.array([[1],[2],[3],[10],[11],[12]])

# Run K-means multiple times with different initializations
for i in range(5):
    kmeans = KMeans(n_clusters=2, init='random', n_init=1, random_state=i)
    kmeans.fit(X)
    print(f"Run {i+1}")
    print("Centroids:", kmeans.cluster_centers_.flatten())
    print("WCSS (Inertia):", kmeans.inertia_)
    print("Labels:", kmeans.labels_)
    print("------")
```
